```python
import torch
import torch.nn as nn
```

```python
class MyModel(nn.Module):
  def __init__(self):
    super().__init__()
    self.fc1 = nn.Linear(10,5)
    self.fc2 = nn.Linear(5,1)
  def forward(self, x):
    x = self.fc1(x)
    x = torch.relu(x)
    x = self.fc2(x)
    return x
model = MyModel()
```

```python
x = torch.rand(1, 10)
output = model(x)
print(output)
```

```python
class Model(nn.Module):
  def __init__(self):
    super().__init__()

    self.fc1 = nn.Linear(10, 64)
    self.fc2 = nn.Linear(64, 32)
    self.fc3 = nn.Linear(32, 1)

  def forward(self, x):
    x = torch.relu(self.fc1(x))
    x = torch.relu(self.fc2(x))
    x = self.fc3(x)
    return x
model = Model()
```

```python
x = torch.rand(1,10)
output = model(x)
```

```python
print(output)
```
